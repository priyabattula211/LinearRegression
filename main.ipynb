```python
import numpy as np
import pandas as pd
data = pd.read_csv(r"C:\Users\priya\Ai_as_skill\House Price Prediction Dataset.csv")
data
```

```python
weight = 0
bias = 0
x_data = data['Area']
y_data = data['Price']
```

```python
def predict(input):
    y_pred = weight * input + bias
    return y_pred
```

```python
def error(y_pred , y_data):
    error = y_pred - y_data
    return error 
def loss(error):
    loss = np.mean(error ** 2)
    return loss
```

```python
def optim(x_data, y_data):
    epochs = range(1000)
    global weight , bias
    learning_rate = 0.01
    n=len(x_data)
    for epoch in epochs:
        pred = predict(x_data)

        errors = error(pred, y_data)

        d_weight = 2/n * np.sum(errors * x_data)
        d_bias = 2/n * np.sum(errors)
        weight = weight - learning_rate * d_weight
        bias = bias - learning_rate * d_bias
        losses = loss(errors)
        print(f"loss: {losses}, epoch: {epoch}")
```

```python
optim(x_data, y_data)
```
